# 🏥 Walk-in Clinic Simulation (SimPy)

See README for the system description and success criteria.

```python
# Starting point: the smallest possible SimPy model — one resource, one process type.
# Run it, understand every line, then build the real clinic on top.
import simpy
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(1)

def patient(env, name, desk):
    arrival = env.now
    with desk.request() as req:
        yield req                                  # wait in line
        yield env.timeout(rng.exponential(3.0))    # service time
    print(f"{name} done at t={env.now:.1f} (waited {env.now - arrival:.1f})")

def arrivals(env, desk):
    i = 0
    while True:
        yield env.timeout(rng.exponential(5.0))    # inter-arrival time
        i += 1
        env.process(patient(env, f"patient_{i}", desk))

env = simpy.Environment()
desk = simpy.Resource(env, capacity=1)
env.process(arrivals(env, desk))
env.run(until=60)
```
